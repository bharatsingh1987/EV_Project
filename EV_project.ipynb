```python
file = "range_vehicle.csv"
range_vehicle = pd.read_csv("/Users/bharatsingh/Downloads/range_vehicle.csv", encoding="ISO-8859-1")
lifetime_km = 250000 # parameter
```

```python
"""Cleaning range_vehicle"""
range_vehicle.columns = (range_vehicle.columns.str.strip().str.lower().str.replace(" ", "_"))
range_vehicle = range_vehicle.rename(columns={
    "fuel_consumption_petrol_(l/100km)": "cons_petrol_l100",
    "fuel_consumption_diesel_(l/100km)": "cons_diesel_l100",
    "energy_consumption_(kwh/100km)": "cons_kwh_100",
    "purchase_price_incl._vat_(eur)": "price_eur",
    "average_range_over_250,000_km_(km)": "avg_range_250000km",})
range_vehicle["mass_(kg)"]=pd.to_numeric(range_vehicle["mass_(kg)"].replace("-", pd.NA),errors="coerce")
range_vehicle["cons_petrol_l100"] = pd.to_numeric(range_vehicle["cons_petrol_l100"].replace("-", pd.NA),errors="coerce")
range_vehicle["cons_diesel_l100"]=pd.to_numeric(range_vehicle["cons_diesel_l100"].replace("-", pd.NA),errors="coerce")
range_vehicle["cons_kwh_100"]=pd.to_numeric(range_vehicle["cons_kwh_100"].replace("-", pd.NA),errors="coerce")
range_vehicle["end_of_life_(tco2e)"]=pd.to_numeric(range_vehicle["end_of_life_(tco2e)"].replace("-", pd.NA),errors="coerce")
range_vehicle["price_eur"]=pd.to_numeric(range_vehicle["price_eur"].replace("-", pd.NA),errors="coerce")
range_vehicle["avg_range_250000km"]=pd.to_numeric(range_vehicle["avg_range_250000km"].replace("-", pd.NA),errors="coerce")
```

```python
countries = pd.read_csv("/Users/bharatsingh/Downloads/mg_price_country_co2e_energy.csv")
countries = countries.rename(columns={
    "average_emissions_intensity_gco2ekwh": "grid_gco2e_kwh",
    "house07station03": "elec_price_eur_kwh",
    "SP95(EUR/L)": "sp95_price_eur_l",
    "Diesel(EUR/L)": "diesel_price_eur_l",
    "Total_CO2_SP95_g_per_L": "co2_petrol_g_l",
    "Total_CO2_Diesel_g_per_L": "co2_diesel_g_l",})
```

```python
#cross join
merged_vehicle_country= range_vehicle.merge(countries, how="cross")
assert len(merged_vehicle_country) == len(range_vehicle)*len(countries)
merged_vehicle_country.to_csv("merged_vehicle_country.csv", index=False)
```

```python
#Quantities of energies on lifetime

km_100 = lifetime_km /100
merged_vehicle_country["petrol_l"] = merged_vehicle_country ["cons_petrol_l100"].fillna(0)*km_100
merged_vehicle_country["diesel_l"] = merged_vehicle_country ["cons_diesel_l100"].fillna(0)*km_100
merged_vehicle_country["elec_kwh"] = merged_vehicle_country["cons_kwh_100"].fillna(0)*km_100

merged_vehicle_country.to_csv("merged_vehicle_country2.csv", index=False)
```

```python
# co2 emited running
merged_vehicle_country["usage_co2e_t"] = (((merged_vehicle_country["petrol_l"]*merged_vehicle_country["co2_petrol_g_l"]) +
                                         (merged_vehicle_country["diesel_l"]*merged_vehicle_country["co2_diesel_g_l"]) +
                                         (merged_vehicle_country["elec_kwh"]*merged_vehicle_country["grid_gco2e_kwh"]))/1000000).round(2) #from grams to tons
```

```python
# Expenses running
merged_vehicle_country["energy_cost"] = ((merged_vehicle_country["petrol_l"]*merged_vehicle_country["sp95_price_eur_l"]) +
                                         (merged_vehicle_country["diesel_l"]*merged_vehicle_country["diesel_price_eur_l"]) +
                                         (merged_vehicle_country["elec_kwh"]*merged_vehicle_country["elec_price_eur_kwh"])).round(2)
```

```python
# lifetime agregate
merged_vehicle_country["lifetime_co2_t"]= (merged_vehicle_country["manufacturing_(tco2e)"]+merged_vehicle_country["end_of_life_(tco2e)"]+merged_vehicle_country["usage_co2e_t"]).round(2)
merged_vehicle_country["lifetime_cost_eur"]= (merged_vehicle_country["price_eur"] + merged_vehicle_country["energy_cost"]).round(2)
merged_vehicle_country["gco2_km"]=(merged_vehicle_country["lifetime_co2_t"]*1000000/(lifetime_km)).round(2)

merged_vehicle_country.to_csv("vehicle_country_lifecycle.csv", index=False)
print(f"{len(merged_vehicle_country)} lignes, {len(merged_vehicle_country.columns)} colonnes")
print(merged_vehicle_country.groupby("energy")["gco2_km"].describe()[["mean", "min", "max"]])
print(merged_vehicle_country.groupby("energy")["price_eur"].describe()[["mean", "min", "max"]])
```

```python
file = "vehicle_country_lifecycle"
range_vehicle = pd.read_csv("vehicle_country_lifecycle.csv")
lifetime_km = 250000 # parameter



from ipywidgets import interact, Dropdown

km = np.linspace(0, lifetime_km, 501)

def show_curves(segment, country):
    sub = merged_vehicle_country[(merged_vehicle_country["segment"] == segment) & (merged_vehicle_country["country"] == country)].set_index("energy")
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    for energy, row in sub.iterrows():
        # CO2: A + x*B + C  (tCO2e) ; A = manufacturing, x = gCO2/km, B= kilometers, C = recyclage
        g_per_km = (row["petrol_l"] * row["co2_petrol_g_l"] + row["diesel_l"] * row["co2_diesel_g_l"]+ row["elec_kwh"] * row["grid_gco2e_kwh"]) / lifetime_km  # g/km
        eol = np.where(km >= lifetime_km, row["end_of_life_(tco2e)"], 0.0)
        ax1.plot(km, row["manufacturing_(tco2e)"] + g_per_km * km / 1e6 + eol, label=energy)

        # Cost: A + x*B (EUR) ; A = cost of buying, x = EUR/km, B= kilometers
        eur_per_km = (row["petrol_l"] * row["sp95_price_eur_l"]+ row["diesel_l"] * row["diesel_price_eur_l"]+ row["elec_kwh"] * row["elec_price_eur_kwh"] ) / lifetime_km
        ax2.plot(km, row["price_eur"] + eur_per_km * km, label=energy)

    ax1.set(title=f"Lifetime CO2 emited- {segment} - {country}", xlabel="km", ylabel="tCO2e")
    ax1.legend()
    ax2.set(title=f"Cumulated cost (Eur) - {segment} - {country}", xlabel="km", ylabel="EUR")
    ax2.legend()
    plt.tight_layout()
    plt.show()

interact(show_curves,segment=Dropdown(options=sorted(merged_vehicle_country["segment"].unique()),description="Segment :"),
    country=Dropdown(options=sorted(merged_vehicle_country["country"].unique()),description="Country :"),)
```
